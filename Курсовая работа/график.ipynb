
$ \begin{cases}
    X_{n+1} = \alpha Y_n(1-\beta Y_n)(1-u_1) \\
    Y_{n+1} = (sX_n + vY_n)(1-u_2)
\end{cases} $

```python
import plotly.express as px
import numpy as np
import pandas as pd
import math
```

```python
n = 100
```

```python
alpha = 0.9 # коэффицент рождаемости
beta = 0.001
u_1 = 0.3 # смертность детей

u_2 = 0.2 # смертность взрослых
s = 0.4 # сколько детей перешло во взрослость
v = 0.2

X_0 = 1000
Y_0 = 500

# защита от "взрыва" значений
THRESH = 1e13
```

```python
X_n = X_0
Y_n = Y_0

end_n = n

X_points = [X_0]
Y_points= [Y_0]
for i in range(n):

    next_X = round(alpha * Y_n * (1 - beta * Y_n) * (1 - u_1))
    next_Y = round((s * X_n + v * Y_n) * (1 - u_2))

    if (abs(next_X) > THRESH or abs(next_Y)) > THRESH:
        end_n = len(X_points)
        break

    X_points.append(next_X)
    Y_points.append(next_Y)

    X_n, Y_n = next_X, next_Y
```

```python
X_points = np.array(X_points)
Y_points = np.array(Y_points)
t = np.arange(len(X_points))
```

```python
print("len X:", len(X_points))
print("len Y:", len(Y_points))
print("end_n:", end_n)
df = pd.DataFrame({
    'n': t,
    "X_n": X_points,
    "Y_n": Y_points
})
fig = px.line(df, x='n', y = ['X_n', 'Y_n'])
fig.show()
```
