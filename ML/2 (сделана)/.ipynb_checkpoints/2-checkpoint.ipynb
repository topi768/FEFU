1. найти изображение и создать для нее маску
2. аргументация.  ( случайно повышать резкость, поворачивать, обрезка и тп) сделать n раз (пусть 100)

```python
import cv2
import numpy as np

# Загрузите изображение
image = cv2.imread('image.jpg')
mask = np.zeros(image.shape[:2], dtype=np.uint8)  # Черный холст

# Нарисуйте фигуры (белым цветом) на маске
cv2.rectangle(mask, (50, 50), (200, 200), 255, -1)  # Прямоугольник
cv2.circle(mask, (300, 300), 100, 255, -1)         # Круг

# Примените маску к изображению
masked_image = cv2.bitwise_and(image, image, mask=mask)

# Сохраните результат
cv2.imwrite('mask.png', mask)
cv2.imwrite('masked_image.jpg', masked_image)
```
