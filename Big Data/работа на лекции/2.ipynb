```python
import vitaldb
import plotly.express as px
```

```python
vf = vitaldb.VitalFile("0001.vital").to_pandas(['Solar8000/BT', 'Solar8000/HR', 'Solar8000/RR_CO2', 'BIS/BIS', 'Solar8000/PLETH_SPO2'], interval=1)
```

```python
vf
vf = vf.reset_index()
vf = vf.rename(columns={"index": 'time',  "Solar8000/BT": "body temperature", 'BIS/BIS': 'Bispectral Index', 'Solar8000/RR_CO2': "Respiratory rate" })
# Solar8000/PLETH_SPO2 - Насыщение крови кислородом
```

```python
vf = vf.dropna()
```

```python
figure = px.line(vf, x="time", y="body temperature")
figure.show()
```
