```python
df['price_bin'] = pd.qcut(
    df['Price'],
    q=5,               # 5 ценовых сегментов
    duplicates='drop'
)
```

```python
df['Weight Capacity (kg)'] = df['Weight Capacity (kg)'].fillna(
    df.groupby(['Compartments', 'price_bin'])['Weight Capacity (kg)']
      .transform('median')
)
```

```python
df['Weight Capacity (kg)'] = df['Weight Capacity (kg)'].fillna(
    df.groupby('Compartments')['Weight Capacity (kg)']
      .transform('median')
)
```

```python
df['Weight Capacity (kg)'] = df['Weight Capacity (kg)'].fillna(
    df['Weight Capacity (kg)'].median().round()
)
```

```python
df = df.drop(columns=['price_bin'])
```
