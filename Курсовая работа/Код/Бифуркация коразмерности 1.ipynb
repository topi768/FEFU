## y = 0.6 v = 0.5 u2 = 0.6



```python
import numpy as np
import plotly.graph_objects as go



# Фиксированные параметры
v = 0.5
u2 = 0.6

array_points = [(0.97, 258.0945), (0.9, 50), (0.97, 50), (0.985, 20)]

u1 = np.linspace(0.01, 0.99, 500)

den = (1 - u1)  * (1 - u2)

# Границы бифуркаций
alpha_tc = (1 - v*(1 - u2)) / den          # q = p - 1
alpha_flip = (1-3 * v *(1-u2)) / den        # q = -1 - p
alpha_ns = (3-2 * v * (1-u2)) / den                        # q = 1

fig = go.Figure()

fig.add_trace(go.Scatter(
    x=u1, y=alpha_tc,
    mode='lines',
    name='Транскритическая (λ = 1)'
))

fig.add_trace(go.Scatter(
    x=u1, y=alpha_flip,
    mode='lines',
    name='Flip (λ = -1)'
))

fig.add_trace(go.Scatter(
    x=u1, y=alpha_ns,
    mode='lines',
    name='Неймарка–Сакера'
))

def create_markers(array_points):
    alphabet = "ABCDE"
    colors = ['#34ebd0', '#3465eb', '#a234eb', '#eb34d6', '#eb3434']
    for i in range(len(array_points)):
        x = array_points[i][0]
        y = array_points[i][1]
        fig.add_trace(go.Scatter(
        x=[x],
        y=[y],
        mode='markers',
        name=alphabet[i],
        marker=dict(
            size=10,
            color=colors[i],
            symbol='circle'
        )
    ))


create_markers(array_points)


fig.update_layout(
    title='Треугольник устойчивости (сечение)',
    xaxis_title='u₁',
    yaxis_title='~α',
    width=800,
    height=500
)

fig.show()
```

## Точка A (u1_point = 0.97, alpha_point = 258.0945)

```python
import pandas as pd
import plotly.express as px

n = 100

v = 0.5
u2 = 0.6

u1 =0.97
tilda_alpha = 258.0945

y0 = 0.6
x0 = 0.3

x_n = x0
y_n = y0
x_points = []
y_points = []
iterations = []

for i in range(n):
    next_x = tilda_alpha * y_n * (1 - y_n)* (1-u1)
    next_y = (x_n + v * y_n)* (1-u2)

    x_points.append(next_x)
    y_points.append(next_y)
    iterations.append(i)

    x_n, y_n = next_x, next_y

x_points = np.array(x_points)
y_points = np.array(y_points)


df = pd.DataFrame({
    "x_n": x_points,
    "y_n": y_points,
    "iteration": iterations
})
fig = px.scatter(
    df,
    x="x_n",
    y="y_n",
    color="iteration",
    color_continuous_scale="Viridis",
)

fig.add_scatter(
    x=df["x_n"],
    y=df["y_n"],
    mode="lines",
    line=dict(color="rgba(200,200,200,0.4)"),
    showlegend=False
)


fig.update_traces(marker=dict(size=6))
fig.show()

```
