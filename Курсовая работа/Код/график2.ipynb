## Рассматриваю точку ff 
$$ (1-u_1)(1-2y)\tilde{\alpha} (1-u_2) = 1  $$
y - стационарная точка, зависящая от $ u_1, u_2, v, \tilde{\alpha} $

```python
import numpy as np
import plotly.graph_objects as go

# Сетки параметров
u1 = np.linspace(0.01, 0.9, 100)
u2 = np.linspace(0.01, 0.9, 100)
U1, U2 = np.meshgrid(u1, u2)

# Поверхность бифуркации ff при v = 0
tilde_alpha = 1 / ((1 - U1) * (1 - U2))

fig = go.Figure(
    data=[go.Surface(
        x=U1,
        y=U2,
        z=tilde_alpha,
        colorscale='Viridis'
    )]
)

fig.update_layout(
    title='поверхность параметров точки (ff) при v = 0',
    scene=dict(
        xaxis_title='u1 (вылов молоди)',
        yaxis_title='u2 (вылов взрослых)',
        zaxis_title='α̃ (воспроизовдство)'
    ),
    width=800,
    height=600
)

fig.show()
```
