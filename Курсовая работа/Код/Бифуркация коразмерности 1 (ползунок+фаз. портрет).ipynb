```python
import numpy as np
import pandas as pd
import plotly.graph_objects as go
import plotly.express as px
import ipywidgets as widgets

import plotly.io as pio
from IPython.display import display, HTML
```

```python
from IPython.display import HTML

HTML("""
<style>
.plot-row {
    display: flex;

}
.plot-row > div {
    flex: 1;
}
</style>
""")
```

```python
from IPython.display import HTML, display

def update_plots(
    u1=0.97,
    tilda_alpha=258,
    v=0.5,
    u2=0.6,
    n=100
):
    # -----------------------------
    # 1. ГРАФИК БИФУРКАЦИЙ
    # -----------------------------
    u1_grid = np.linspace(0.01, 0.99, 500)
    den = (1 - u1_grid) * (1 - u2)

    alpha_tc = (1 - v * (1 - u2)) / den
    alpha_flip = (1 - 3 * v * (1 - u2)) / den
    alpha_ns = (3 - 2 * v * (1 - u2)) / den

    fig1 = go.Figure()

    fig1.add_trace(go.Scatter(
        x=u1_grid, y=alpha_tc,
        mode='lines',
        name='Транскритическая (λ = 1)'
    ))

    fig1.add_trace(go.Scatter(
        x=u1_grid, y=alpha_flip,
        mode='lines',
        name='Flip (λ = -1)'
    ))

    fig1.add_trace(go.Scatter(
        x=u1_grid, y=alpha_ns,
        mode='lines',
        name='Неймарка–Сакера'
    ))

    fig1.add_trace(go.Scatter(
        x=[u1],
        y=[tilda_alpha],
        mode='markers',
        name='Текущие параметры',
        marker=dict(size=12, color='red')
    ))

    fig1.update_layout(
        title='Треугольник устойчивости (сечение)',
        xaxis_title='u₁',
        yaxis_title='~α',
        width=700,
        height=450
    )

    # -----------------------------
    # 2. ФАЗОВЫЙ ПОРТРЕТ
    # -----------------------------
    x0, y0 = 0.3, 0.6
    x_n, y_n = x0, y0

    xs, ys, it = [], [], []

    for i in range(n):
        next_x = tilda_alpha * y_n * (1 - y_n) * (1 - u1)
        next_y = (x_n + v * y_n) * (1 - u2)

        xs.append(next_x)
        ys.append(next_y)
        it.append(i)

        x_n, y_n = next_x, next_y

    df = pd.DataFrame({
        "x": xs,
        "y": ys,
        "iteration": it
    })

    fig2 = px.scatter(
        df,
        x="x",
        y="y",
        color="iteration",
        color_continuous_scale="Viridis",
        title="Фазовый портрет"
    )

    fig2.add_scatter(
        x=df["x"],
        y=df["y"],
        mode="lines",
        line=dict(color="rgba(180,180,180,0.4)"),
        showlegend=False
    )

    fig2.update_traces(marker=dict(size=6))
    fig2.update_layout(width=700, height=450)

    # -----------------------------
    # ГОРИЗОНТАЛЬНЫЙ ВЫВОД
    # -----------------------------
    html = f"""
    <div class="plot-row">
        <div class="plot">{fig1.to_html(include_plotlyjs=False)}</div>
        <div class="plot">{fig2.to_html(include_plotlyjs=False)}</div>
    </div>
    """

    display(HTML(html))
```

```python
widgets.interact(
    update_plots,
    u1=widgets.FloatSlider(min=0.01, max=0.99, step=0.001, value=0.97),
    tilda_alpha=widgets.FloatSlider(min=0, max=400, step=1, value=258),
    v=widgets.FloatSlider(min=0.01, max=0.99, step=0.001, value=0.5),
    u2=widgets.FloatSlider(min=0.01, max=0.99, step=0.001, value=0.6),
    n=widgets.IntSlider(min=1, max=300, step=1, value=100),
)
```
