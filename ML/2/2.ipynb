1. найти изображение и создать для нее маску
2. аугментация (случайно повышать резкость, поворачивать, обрезка и тп) сделать n раз (пусть 100)

```python
import cv2
import numpy as np

# Загрузить изображение
image = cv2.imread('2.jpg')
h, w = image.shape[:2]

# Создать черную маску
mask = np.zeros((h, w), dtype=np.uint8)

x1 = 550
y1 = 20
x2 = 900
y2 = 330

# Нарисовать белый прямоугольник на маске
cv2.rectangle(mask, (x1, y1), (x2, y2), 255, -1)

# Применить маску к изображению
masked_image = cv2.bitwise_and(image, image, mask=mask)

# СОХРАНИТЬ РЕЗУЛЬТАТ
cv2.imwrite('masked_result.jpg', masked_image)

print("Изображение сохранено как 'masked_result.jpg'")

# Показать результат (опционально)
# cv2.imshow('Masked Image', masked_image)
# cv2.waitKey(0)
# cv2.destroyAllWindows()
```
