```python
import re
import regex
from nltk.stem.snowball import SnowballStemmer
from collections import Counter

with open("Bible_txt.txt", "r", encoding="cp1251") as f:
    text = f.read()

emoji_re = regex.compile(r"[\p{So}\p{Cn}]+", flags=regex.UNICODE)
non_word_re = regex.compile(r"[^0-9\p{L}'\s]+", flags=regex.UNICODE)
multispace_re = re.compile(r"\s+")

stemmer = SnowballStemmer("russian")


def clean_text(text):
    # A) Приведение к нижнему регистру
    text = text.lower()

    # Б) Удаление знаков препинания
    text = non_word_re.sub(" ", text)

    # В) Удаление лишних пробелов
    text = multispace_re.sub(" ", text).strip()

    # Г) Удаление мусора
    text = emoji_re.sub(" ", text)
    return text

def tokenize_regex(text: str):
    tokens = regex.findall(r"\p{L}[\p{L}']*", text, flags=regex.UNICODE)
    return tokens



def stem_tokens(tokens):
    # Д) Стемминг
    return [stemmer.stem(t) for t in tokens]


# text = ("Привет!!  Это   Тестовая   СТРОКА 😊 <b>проверка</b>  — коты, кота и КОТИКИ "
#             "любят бегать!!!  😎")

cleaned = clean_text(text)
tokens = tokenize_regex(cleaned)
stems = stem_tokens(tokens)
#
# print("cleaned:", cleaned)
# print("stems:", stems)
bow = Counter(tokens)
print(bow.most_common(50)  )

```
