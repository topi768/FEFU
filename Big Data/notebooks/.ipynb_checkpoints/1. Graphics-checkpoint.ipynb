```python
import plotly.express as px

df['Sex_ru'] = df['Sex'].map({'Female': 'Женский', 'Male': 'Мужской'})
color_map = {'Женский': '#e051a0', 'Мужской': 'blue'}

fig = px.histogram(
    df,
    x="Sex_ru",
    y="HeightInMeters",
    color="Sex_ru",
    color_discrete_map=color_map,
    barmode="overlay",
    histfunc="avg",
    title="Средний рост по полу",
    height=500,
    width=600,
    labels={"Sex_ru": "Пол"},
    opacity=0.8,
    hover_data={"HeightInMeters": False},
)
fig.update_traces(
    hovertemplate="<b>%{x}</b><br>Средний рост: %{y:.2f} м<extra></extra>",
    opacity=0.7
)
fig.update_yaxes(title_text="Рост (м)", dtick=0.2)

fig.show()
```

Средний рост мужчин - почти метр 1.8. У женщин 1.6

## 2. Инфаркт по штатам

```python
df['HeartAttackCount'] = df['HadHeartAttack'].map({'Yes': 1, 'No': 0})


heart_attack_by_state = df.groupby("State")["HeartAttackCount"].sum().reset_index()

heart_attack_by_state = heart_attack_by_state.sort_values(by="HeartAttackCount", ascending=False)

print(heart_attack_by_state.head(3))
```

```python
population = pd.read_csv("../data/raw/us_pop_by_state.csv")
population = population.drop('rank', axis=1)
population = population.drop('percent_of_total', axis=1)
population = population.drop('state', axis=1)

population.head(3)
```

```python
import pandas as pd
import plotly.graph_objects as go


population.rename(columns={'2020_census': 'Population'}, inplace=True)

state_to_code = {
    'Alabama': 'AL', 'Alaska': 'AK', 'Arizona': 'AZ', 'Arkansas': 'AR', 'California': 'CA',
    'Colorado': 'CO', 'Connecticut': 'CT', 'Delaware': 'DE', 'Florida': 'FL', 'Georgia': 'GA',
    'Hawaii': 'HI', 'Idaho': 'ID', 'Illinois': 'IL', 'Indiana': 'IN', 'Iowa': 'IA', 'Kansas': 'KS',
    'Kentucky': 'KY', 'Louisiana': 'LA', 'Maine': 'ME', 'Maryland': 'MD', 'Massachusetts': 'MA',
    'Michigan': 'MI', 'Minnesota': 'MN', 'Mississippi': 'MS', 'Missouri': 'MO', 'Montana': 'MT',
    'Nebraska': 'NE', 'Nevada': 'NV', 'New Hampshire': 'NH', 'New Jersey': 'NJ', 'New Mexico': 'NM',
    'New York': 'NY', 'North Carolina': 'NC', 'North Dakota': 'ND', 'Ohio': 'OH', 'Oklahoma': 'OK',
    'Oregon': 'OR', 'Pennsylvania': 'PA', 'Rhode Island': 'RI', 'South Carolina': 'SC', 'South Dakota': 'SD',
    'Tennessee': 'TN', 'Texas': 'TX', 'Utah': 'UT', 'Vermont': 'VT', 'Virginia': 'VA',
    'Washington': 'WA', 'West Virginia': 'WV', 'Wisconsin': 'WI', 'Wyoming': 'WY',
    'District of Columbia': 'DC', 'Puerto Rico': 'PR', 'Guam': 'GU', 'Virgin Islands': 'VI'
}
merged_df = heart_attack_by_state.copy()
merged_df['code'] = merged_df['State'].map(state_to_code)
merged_df.dropna(subset=['code'], inplace=True)


final_df = pd.merge(merged_df, population, left_on='code', right_on='state_code')

final_df['HeartAttackRate'] = (final_df['HeartAttackCount'] / final_df['Population']) * 100000


final_df['text'] = final_df['State'].map(russian_names) + '<br>' + \
    'Случаев на 100 тыс. чел.: ' + final_df['HeartAttackRate'].round(2).astype(str) + '<br>' + \
    'Всего случаев: ' + final_df['HeartAttackCount'].astype(str)


fig = go.Figure(data=go.Choropleth(
    locations=final_df['code'],
    z=final_df['HeartAttackRate'],
    locationmode='USA-states',
    colorscale='Reds',
    autocolorscale=False,
    text=final_df['text'],
    marker_line_color='white',
    colorbar=dict(
        title=dict(
            text="Случаев на 100 тыс. чел."
            )
    )))

fig.update_layout(
    title_text='Частота случаев инфаркта на 100.000 населения по штатам США',
    geo = dict(
        scope='usa',
        projection=go.layout.geo.Projection(type = 'albers usa'),
        showlakes=True,
        lakecolor='rgb(255, 255, 255)'),
    margin={"r":0,"t":50,"l":0,"b":0}
)

fig.show()
```

Печальная ситуация в центрально - северных штатах и в Вермонте - у них больше всего людей с инфарктом на 100к населения.

## 3. Какой изъян больше всего коррелирует с наличием инфаркта

```python
import pandas as pd
import plotly.express as px

features = ['HadAsthma', 'HadSkinCancer', 'HadDepressiveDisorder',
            'HadArthritis', 'DifficultyWalking',
            'AlcoholDrinkers', 'DifficultyConcentrating']

feature_ru = {
    'HadAsthma': 'Астма',
    'DifficultyConcentrating': 'Трудности с концентрацией',
    'HadSkinCancer': 'Рак кожи',
    'HadDepressiveDisorder': 'Депрессия',
    'HadArthritis': 'Артрит',
    'DifficultyWalking': 'Проблемы с ходьбой',
    'SmokerStatus': 'Курильщик',
    'AlcoholDrinkers': 'Пьет алкоголь',
    'PneumoVaxEver': 'Прививка от пневмонии',
}

df_heart_attack = df[df['HadHeartAttack'] == 'Yes']

# Считаем долю людей с инфарктом среди тех, у кого есть каждый признак
ratios = {}
for feature in features:
    total_with_feature = (df[feature] == 'Yes').sum()
    with_heart_attack = (df_heart_attack[feature] == 'Yes').sum()
    if total_with_feature > 0:
        ratios[feature] = with_heart_attack / total_with_feature
    else:
        ratios[feature] = 0

df_ratios = pd.DataFrame({
    'Feature': [feature_ru[f] for f in ratios.keys()],
    'Proportion': list(ratios.values())
})

fig = px.bar(df_ratios, x='Feature', y='Proportion',
             text=df_ratios['Proportion'].apply(lambda x: f"{x:.2%}"),
             title='Доля людей с инфарктом среди имеющих указанный признак',
             height=500, width=800)

fig.update_traces(textposition='outside', textfont_size=12)
fig.update_yaxes(range=[0, df_ratios['Proportion'].max() * 1.15],
                 tickformat=".0%")
fig.update_layout(xaxis_title="Признаки", yaxis_title="Доля с инфарктом (%)")
fig.show()
```
