```python
df['Crime List'] = df['Type of the Crime'].str.split(',')
```

```python
df['Crime List'] = df['Crime List'].apply(lambda lst: [x.strip() for x in lst])
mlb = pd.get_dummies(df['Crime List'].explode()).groupby(level=0).sum()

df = pd.concat([df.drop(columns=['Crime List']), mlb], axis=1)
```

```python
import re
import numpy as np

def punishment_to_years(x):
    if pd.isna(x):
        return np.nan
    x = x.lower().strip()

    if '-' in x:
        nums = re.findall(r'\d+\.?\d*', x)
        nums = [float(n) for n in nums]
        return np.mean(nums)

    if 'month' in x:
        num = float(re.findall(r'\d+\.?\d*', x)[0])
        return num / 12

    if 'year' in x:
        num = float(re.findall(r'\d+\.?\d*', x)[0])
        return num

    if 'death' in x:
        return 100
    if 'life' in x:
        return 75

    # особые случаи
    if 'community service' in x:
        return 1
    if 'probation' in x or 'not sentenced' in x or 'present' in x:
        return 0

    return np.nan
```

```python
df['Punishment for a crime'] = df['Punishment for a crime'].apply(punishment_to_years)
```
