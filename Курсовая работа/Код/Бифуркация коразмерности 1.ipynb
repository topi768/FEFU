## Фазовый портрет с множеством точек

```python
import numpy as np
import plotly.graph_objects as go

# Параметры системы
v = 0.5
u1 = 0.97
u2 = 0.6
tilda_alpha = 258.0945

n_iter = 200     # всего итераций
burn_in = 20     # количество итераций для обрезки (transient)

# Сетка начальных точек
x0_values = np.linspace(0.1, 0.5, 5)
y0_values = np.linspace(0.1, 0.7, 5)

fig = go.Figure()

for x0 in x0_values:
    for y0 in y0_values:
        x_n, y_n = x0, y0
        xs, ys = [], []

        for i in range(n_iter):
            x_next = tilda_alpha * y_n * (1 - y_n) * (1 - u1)
            y_next = (x_n + v * y_n) * (1 - u2)

            x_n, y_n = x_next, y_next

            # Сохраняем только после "burn-in"
            if i >= burn_in:
                xs.append(x_n)
                ys.append(y_n)

        fig.add_trace(go.Scatter(
            x=xs, y=ys,
            mode='lines+markers',
            marker=dict(size=4),
            line=dict(width=1, color='rgba(150,150,150,0.6 (сдана))'),
            showlegend=False
        ))

fig.update_layout(
    title='Фазовый портрет (с обрезкой первых 20 итераций)',
    xaxis_title='x',
    yaxis_title='y',
    width=700,
    height=600
)

fig.show()
```

## Карта режимов

```python
import numpy as np
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap, BoundaryNorm

def detect_period(arr, max_period=20, tol=1e-6):

    arr = np.array(arr, dtype=float)

    if not np.all(np.isfinite(arr)):
        return None

    arr = np.round(arr, int(-np.log10(tol)))

    n = len(arr)

    for p in range(1, min(max_period, n // 2) + 1):
        is_periodic = True
        for i in range(n - p):
            if arr[i] != arr[i + p]:
                is_periodic = False
                break
        if is_periodic:
            return p

    return None
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Фиксированные параметры

u1 = 0.97
u2 = 0.6


mu2 = 0.6
v = 0.5
# Диапазоны параметров
u1_vals = np.linspace(-0.9, 0.9, 900)
tilda_alpha_vals = np.linspace(0.1, 400, 400)

N_trans = 150   # число итераций
N_iter = 20     # - столько последних чисел сохраняем, чтобы понять структуру

# Начальные условия
x0, y0 = 0.3, 0.6





regime_map = np.full((len(tilda_alpha_vals), len(u1_vals)), -1)

for i, alpha_tilde in enumerate(tilda_alpha_vals):

    for j, u1 in enumerate(u1_vals):
        x, y = x0, y0

        try:
            # Прогрев
            for _ in range(N_trans):
                x_new = alpha_tilde * (1 - u1) * y * (1 - y)
                y_new = (x + v * y) * (1 - u2)
                x, y = x_new, y_new
                if not np.isfinite(x) or not np.isfinite(y):
                    raise OverflowError

            # Сохраняем последние значения y
            y_vals = []
            for _ in range(N_iter):
                x_new = alpha_tilde * (1 - u1) * y * (1 - y)
                y_new = (x + v * y) * (1 - u2)
                x, y = x_new, y_new
                if not np.isfinite(x) or not np.isfinite(y):
                    raise OverflowError
                y_vals.append(y)

                p = detect_period(y_vals, max_period=20)

                if p is None:
                    regime_map[i, j] = 2      # хаос
                elif p == 1:
                    regime_map[i, j] = 0      # период-1
                elif p == 2:
                    regime_map[i, j] = 1      # период-2
                else:
                    regime_map[i, j] = 2      # период > 2
        except OverflowError:
            regime_map[i, j] = 3  # не удалось посчитать. Либо nan либо inf → белый цвет


cmap = ListedColormap(['#84fc03', '#03befc', 'red', '#000000'])
bounds = [-0.5, 0.5, 1.5, 2.5, 3.5]  # границы для нормировки
norm = BoundaryNorm(bounds, cmap.N)

plt.imshow(
    regime_map,
    extent=[u1_vals[0], u1_vals[-1], tilda_alpha_vals[0], tilda_alpha_vals[-1]],
    origin='lower',
    aspect='auto',
    cmap=cmap,
    norm=norm
)

cbar = plt.colorbar(ticks=[0,1,2, 3])
cbar.ax.set_yticklabels(['1 период', '2 период', '>2 период', 'не удалось посчитать'])
```
