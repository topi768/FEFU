```python
import vitaldb
import pandas as pd
import plotly.express as px
from scipy.fft import fft, fftfreq
import numpy as np
```

```python
# vitaldb.vital_trks('0001.vital')
```

```python
vf = vitaldb.VitalFile("0001.vital").to_pandas(['SNUADC/ECG_V5'], interval=1/500)
```

```python
vf = vf.reset_index()
vf = vf.rename(columns={"index": 'time' })
vf = vf.dropna()
```

```python
seconds_to_analyze = 10  # анализируем только 10 секунд
samples_per_second = 500
max_samples = seconds_to_analyze * samples_per_second
vf = vf.head(max_samples)
```

```python
figure = px.line(vf, x="time", y="SNUADC/ECG_V5")
figure.show()
```
