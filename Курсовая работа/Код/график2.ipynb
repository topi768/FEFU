```python
import numpy as np
import plotly.express as px
import pandas as pd

# Параметры
y = 0.6
v0 = 0.5
u0 = 0.5

# Массив u1
u1 = np.linspace(0, 0.99, 500)  # не включаем 1, чтобы избежать деления на ноль

# Вычисляем tilde_alpha
tilde_alpha = - (v0*(1-u0) - 1) / ((1-u0)*(1-u1)*(1 - 2*y))

# Создаем DataFrame для plotly
df = pd.DataFrame({'u1': u1, 'tilde_alpha': tilde_alpha})

# Строим интерактивный график
fig = px.line(df, x='u1', y='tilde_alpha', title='Граница устойчивости (транскритическая бифуркация)',
              labels={'u1': 'u1 — доля вылова малых рыб', 'tilde_alpha': 'tilde_alpha — рождаемость и переход к взрослым'})

fig.update_layout(template='plotly_white')
fig.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Параметры
y = 0.6
v = 0.9

# Сетки для u1 и u2, избегаем 1
u1 = np.linspace(0, 0.999, 100)
u2 = np.linspace(0, 0.999, 100)
U1, U2 = np.meshgrid(u1, u2)

# Вычисляем tilde_alpha
tilde_alpha = (v*(1-U2) - 1) / ((1-U1)*(1-U2)*(1-2*y))

# Создаём 3D график
fig = plt.figure(figsize=(10,7))
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(U1, U2, tilde_alpha, cmap='viridis')
ax.set_xlabel('u1 (доля вылова малых рыб)')
ax.set_ylabel('u2 (доля вылова взрослых рыб)')
ax.set_zlabel('tilde_alpha')
ax.set_title('3D график tilde_alpha(u1, u2)')

plt.show()
```
